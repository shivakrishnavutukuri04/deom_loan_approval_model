```python
import numpy as np
import pandas as pd
import sklearn
```

## step 1 - split the data into X(inputs) and y(output)

```python
X = df.drop('LOAN_APPROVAL',axis=1)
```

```python
y = df[['LOAN_APPROVAL']]
```

## step 2 - split the data into train and test (80-20)

```python
from sklearn.model_selection import train_test_split

X_train , X_test , y_train , y_test  = train_test_split(X,y,train_size=0.8,random_state=4)
```

```python
## step 3 - Feature engineering on Training Data ( X_train and y_train ) 
```

```python
X_train
```

```python
cat_df = X_train.select_dtypes(include = 'object')
```

```python
cat_df 
```

```python
from sklearn.preprocessing import OneHotEncoder

OHE = OneHotEncoder(sparse_output=False).set_output(transform='pandas')

cat_df_trans = OHE.fit_transform(cat_df)
```

```python
cat_df_trans
```

```python
OHE.categories_
```

```python
num_df = X_train[['age','balance','day','duration','campaign','pdays']]
```

```python
no_trans = X_train[['default','housing_loan','personal_loan','previous']]
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler().set_output(transform='pandas')

num_df_trans = scaler.fit_transform(num_df)

```

```python
num_df_trans
```

```python
num_df_trans.reset_index(inplace=True)
```

```python
cat_df_trans.reset_index(inplace=True)
```

```python
no_trans.reset_index(inplace=True)
```

```python
X_train_trans = pd.concat([num_df_trans,cat_df_trans,no_trans],axis=1)
```

```python
X_train_trans
```

```python
X_train_trans.pop('index')
```

```python
X_train_trans
```

```python
y_train.reset_index(inplace=True)
```

```python
y_train.pop('index')
```

```python
y_train
```

## step 4 --> Training model using algorithm 

```python
from sklearn.neighbors import KNeighborsClassifier

model = KNeighborsClassifier()

model.fit(X_train_trans,y_train)
```

## step 5 -- Feature Engineering on Test Data 

```python
X_test
```

```python
num_df = X_test[['age','balance','day','duration','campaign','pdays']]
```

```python
cat_df = X_test.select_dtypes(include='object')
```

```python
no_trans = X_test[['default','housing_loan','personal_loan','previous']]
```

```python
## num trans
num_df_trans = scaler.transform(num_df)
num_df_trans
```

```python
## cat_df transformation 

cat_df_trans = OHE.transform(cat_df)
cat_df_trans
```

```python
num_df_trans.reset_index(inplace=True)
```

```python
cat_df_trans.reset_index(inplace=True)
```

```python
no_trans.reset_index(inplace=True)
```

```python
X_test_trans = pd.concat([num_df_trans,cat_df_trans,no_trans],axis=1)
```

```python
X_test_trans
```

```python
X_test_trans.pop('index')
```

```python
y_test.reset_index(inplace=True)
```

```python
y_test.pop('index')
```

## step 6 --> predicting on X_test_trans

```python
y_pred = model.predict(X_test_trans)
y_pred
```

## step 7 -- passing y_test(actual) and y_pred(model_prediction) to Evaluation Metric

```python
from sklearn.metrics import accuracy_score

accuracy_score(y_test,y_pred)
```
